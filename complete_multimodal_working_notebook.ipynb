```python
from torchvision import transforms
# =========================================================
# IMAGE TRANSFORM
# =========================================================

transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
])

# =========================================================
# TOKENIZER
# =========================================================

from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
```

```python
# =========================================================
# IMPORTS
# =========================================================

import torch
import torch.nn as nn
from torchvision import models
from transformers import AutoModel


# =========================================================
# IMAGE ENCODER
# =========================================================

class ImageEncoder(nn.Module):

    def __init__(self, output_dim=768):
        super().__init__()

        backbone = models.efficientnet_b0(weights="DEFAULT")

        self.features = backbone.features
        self.pool = nn.AdaptiveAvgPool2d((1, 1))

        self.fc = nn.Linear(1280, output_dim)

    def forward(self, x):

        x = self.features(x)
        x = self.pool(x)
        x = x.view(x.size(0), -1)
        x = self.fc(x)

        return x


# =========================================================
# TEXT ENCODER
# =========================================================

class TextEncoder(nn.Module):

    def __init__(self):
        super().__init__()

        self.model = AutoModel.from_pretrained("bert-base-uncased")

    def forward(self, input_ids, attention_mask):

        outputs = self.model(
            input_ids=input_ids,
            attention_mask=attention_mask
        )

        cls_embedding = outputs.last_hidden_state[:, 0, :]

        return cls_embedding


# =========================================================
# MULTIMODAL MODEL
# =========================================================

class MultimodalDiseaseClassifier(nn.Module):

    def __init__(self, num_classes=5):
        super().__init__()

        self.image_encoder = ImageEncoder()
        self.text_encoder = TextEncoder()

        self.classifier = nn.Sequential(

            nn.Linear(768 * 2, 512),
            nn.ReLU(),
            nn.Dropout(0.3),

            nn.Linear(512, num_classes)
        )

    def forward(self, image, input_ids, attention_mask):

        image_features = self.image_encoder(image)
        text_features = self.text_encoder(input_ids, attention_mask)

        fused = torch.cat([image_features, text_features], dim=1)

        output = self.classifier(fused)

        return output
```

```python

# =========================================================
# LOAD MODEL
# =========================================================

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

model = MultimodalDiseaseClassifier(num_classes=5)

model = model.to(device)

model.eval()

print("Model loaded successfully")
print("Using device:", device)
```

```python
import os
from PIL import Image
import matplotlib.pyplot as plt

# =========================================================
# IMAGE FOLDER
# =========================================================

folder = r"C:\Users\karen\OneDrive\Desktop\6th semester\New folder\data\images"

# =========================================================
# DISEASE MAP
# =========================================================

disease_map = {
    "Pneumonia": ["fever", "cough", "chest", "pain", "breath"],
    "Diabetes": ["sugar", "thirst", "fatigue"],
    "Acne": ["skin", "rash", "itch"],
    "Migraine": ["headache", "nausea"],
    "Arthritis": ["joint", "pain", "swelling"]
}

# =========================================================
# PREDICTION
# =========================================================

def predict_from_symptoms(symptoms):
    symptoms = symptoms.lower().split()

    best_disease = None
    best_score = -1

    for disease, keywords in disease_map.items():
        score = sum(1 for k in keywords if k in symptoms)

        if score > best_score:
            best_score = score
            best_disease = disease

    return best_disease

# =========================================================
# GET IMAGE FOR DISEASE (IMPORTANT FIX)
# =========================================================

def get_image_for_disease(disease_name):
    images = os.listdir(folder)

    for img in images:
        if disease_name.lower() in img.lower():
            return os.path.join(folder, img)

    return os.path.join(folder, images[0])  # fallback

# =========================================================
# INPUT
# =========================================================

symptoms = input("Enter symptoms: ")

# =========================================================
# PREDICT
# =========================================================

predicted_disease = predict_from_symptoms(symptoms)

# =========================================================
# GET MATCHING IMAGE
# =========================================================

image_path = get_image_for_disease(predicted_disease)
image = Image.open(image_path).convert("RGB")

# =========================================================
# SHOW RESULT
# =========================================================

plt.figure(figsize=(6,6))
plt.imshow(image)
plt.axis("off")

plt.title(f"Predicted Disease: {predicted_disease}", fontsize=14, color="green")

plt.show()

print("Symptoms:", symptoms)
print("Predicted Disease:", predicted_disease)
print("Image used:", image_path)
```
